```python
class calculator :

    # Constructor accepts 2 required arguments and 2 optional arguments.
    # c and d have default values of 0, so they don't have to be provided.
    def __init__ (self , a , b , c = 0 , d = 0) :
        self.a = a 
        self.b = b 
        self.c = c 
        self.d = d 

        # Call the add() method when the object is created.
        self.add()

    # The same add() method can perform addition with
    # different numbers of values depending on the arguments provided.
    def add (self) :

        # If only a and b are provided, add only two numbers.
        if(self.c == 0 & self.d == 0) :
            print(self.a + self.b)

        # If c is provided but d is not provided,
        # add three numbers.
        elif(self.d == 0):
            print(self.a + self.b + self.c) 

        # If all four values are provided,
        # add all four numbers.
        else :
            print(self.a + self.b + self.c + self.d) 
```

```python
c1 = calculator(20 , 20)
```

```python
c2 = calculator(20 , 20 , 20)
```

```python
c3 = calculator(20 , 20 , 20 , 20)
```
