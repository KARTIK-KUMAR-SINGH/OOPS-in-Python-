```python
# Parent class 1
# Stores and provides information related to the father.
class father :

    def __init__ (self , f_name , f_occupation): 
        self.f_name = f_name
        self.f_occupation = f_occupation

    # Method specific to the Father class
    def show(self):
        return f"The father name is :- {self.f_name} and his occupation is :- {self.f_occupation}"


# Parent class 2
# Stores and provides information related to the mother.
class mother :
    def __init__ (self , m_name , m_occupation) :
        self.m_name = m_name
        self.m_occupation = m_occupation

    def show(self) :
        return f"The mother name is :- {self.m_name} and her occupation is :- {self.m_occupation}"

# Both Father and Mother define a method named show().
# Child inherits from both classes.
class child(father , mother) :
    def __init__ (self , f_name , f_occupation , m_name , m_occupation , c_name , c_age) :
        self.c_name = c_name
        self.c_age = c_age

        # Initialize both parent classes.
        father.__init__(self , f_name , f_occupation)
        mother.__init__(self , m_name , m_occupation)

    def child_intro(self):
        return f"The child name is :- {self.c_name} and his age is :- {self.c_age}"
```

```python
c1 = child("ramesh" , "Engineer" , "sunita" , "teacher" , "raj" , 21)
```

```python
# Since Child inherits from Father first, Python's MRO is:
# Child -> Father -> Mother -> object
#
# Both Mother and Father have a show() method.
# Python searches according to the MRO and finds Mother's show() first.
print(c1.show())
print(c1.child_intro())
# Display the Method Resolution Order (MRO) of Child.
# Python searches for methods in this order.
print(child.mro())
```
