```python
class student :

    # Constructor to initialize the data of each student object
    def __init__ (self , name , age) :
        self.name = name 
        self.age = age

    # Static method does not use self or cls.
    # It can be called directly using the class name without creating an object.
    @staticmethod
    def is_adult(age) :

        # Check whether the given age is 18 or above
        if(age >= 18) :
            return f"adult , since the age is  :- {age}"
        else :
            return f"not adult , since the age is :- {age}"
```

```python
# Calling the static method directly using the class name.
# No student object is required to call this method.
print(student.is_adult(21))
print(student.is_adult(17))
```
