```python
class student :
    school = "ABC school"

    def __init__ (self , name , roll_no , marks) :
        self.name  = name 
        self.roll_no = roll_no
        self.marks = marks
```

```python
s1 = student('kartik' , 60 , 94)
s2 = student('abhishek' , 62 , 96)

print(s1.school)
print(s1.name)
print(s1.roll_no)
print(s1.marks)

print(s2.school)
print(s2.name)
print(s2.roll_no)
print(s2.marks)
```
