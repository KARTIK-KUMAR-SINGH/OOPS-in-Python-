```python
# Creating a Student class
class student :

    # This is a class attribute.
    # It belongs to the class and is shared by all instances.
    school = "ABC school"

    # __init__ is the constructor.
    # It runs automatically when an instance is created.
    def __init__ (self , name , roll_no , marks) :

        # These are instance attributes.
        # Each instance has its own name, roll_no and marks.
        self.name  = name 
        self.roll_no = roll_no
        self.marks = marks
```

```python
# Creating the first instance of the Student class
s1 = student('kartik' , 60 , 94)

# Creating the second instance of the Student class
s2 = student('abhishek' , 62 , 96)

# Accessing the class attribute using the first instance
# Both s1 and s2 can access the same class attribute.
print(s1.school)

# Accessing instance attributes of s1
print(s1.name)
print(s1.roll_no)
print(s1.marks)

# Accessing the class attribute using the second instance
print(s2.school)

# Accessing instance attributes of s2
print(s2.name)
print(s2.roll_no)
print(s2.marks)
```
