```python
class person :
    def __init__ (self , name , age) :
        self.name = name
        self.age = age

    def introduce(self) :
        return f"the person name is  :- {self.name} and this age is :- {self.age}"


class student(person) :
    # Student defines its own constructor to add student-specific data.
    def __init__(self , name , age , roll_no) :
        # Initialize the attributes defined in the parent class.
        # super() calls Person.__init__().

        # Student-specific attribute.
        self.roll_no = roll_no 
        super().__init__(name , age)

    # Student has its own method in addition to inherited methods.
    def student_info(self) :
        return f"The student roll no is :- {self.roll_no}"
```

```python
# Creating a Student object.
# name and age are handled by the parent class,
# while roll_no is handled by the Student class.
s1 = student('abhishek' , 21 , 62)

# Calling the method inherited from Person.
print(s1.introduce())

# Calling the method defined specifically in Student.
print(s1.student_info())
```
