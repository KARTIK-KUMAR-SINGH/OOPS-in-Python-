```python
# Student inherits from Person.
# Since Student has no __init__ or introduce() of its own,
# it automatically uses the methods inherited from Person.
class person :
    def __init__ (self , name , age) :
        self.name = name
        self.age = age

    def introduce(self) :
        return f"the person name is  :- {self.name} and this age is :- {self.age}"

class student(person) :
    pass
```

```python
# Creating a Student object uses the inherited Person.__init__()
# and introduce() method.
s1 = student("abhishek" , 20)
# Calling the inherited introduce() method.
s1.introduce()
```
