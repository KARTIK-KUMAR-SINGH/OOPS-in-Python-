```python
class calculator :

    # Constructor used to initialize the value of the object
    def __init__ (self , a) :
        self.a = a 

    # __str__ is called automatically when we print the object.
    # It defines how the object should be represented as a string.
    def __str__ (self) :
        return f"The values are a = {self.a}"

    # __add__ overloads the '+' operator for calculator objects.
    # self represents the left object and other represents the right object.
    def __add__ (self , other) :
        return calculator(self.a + other.a)

    # __sub__ overloads the '-' operator for calculator objects.
    def __sub__ (self , other) :
        return calculator(self.a - other.a)

    # __mul__ overloads the '*' operator for calculator objects.
    def __mul__ (self , other):
        return calculator(self.a * other.a)

    # __truediv__ overloads the '/' operator for calculator objects.
    def __truediv__ (self , other):
        return calculator(self.a / other.a)
```

```python
# Creating a calculator object with value 10
c1 = calculator(10)
print(c1)
```

```python
# Normal addition happens before the object is created.
# Here 10 + 40 = 50, then calculator(50) is created.
c2 = calculator(10 + 40)
print(c2)
```

```python
# Normal subtraction happens before the object is created.
# Here 40 - 20 = 20, then calculator(20) is created.c3 = calculator(40 - 20)
print(c3)
```

```python
# Normal multiplication happens before the object is created.
# Here 20 * 40 = 800, then calculator(800) is created.
c4 = calculator(20 * 40)
print(c4)
```

```python
# Normal division happens before the object is created.
# Here 40 / 20 = 2.0, then calculator(2.0) is created.
c5 = calculator(40 / 20)
print(c5)
```
