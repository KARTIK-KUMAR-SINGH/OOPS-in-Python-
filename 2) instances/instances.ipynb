```python
# Creating a Student class
class student :
    # __init__ is the constructor.
    # It runs automatically whenever we create a new instance.
    def __init__ (self , name , roll_no , marks) :

        # These are instance attributes.
        # Each instance gets its own name, roll_no and marks.
        self.name  = name 
        self.roll_no = roll_no
        self.marks = marks
```

```python
# Creating two instances of the Student class
s1 = student('abhishek' , 60 , 100)
s2 = student('ankit' , 80 , 100)


# Accessing the attributes of instance s1
print("Instance 1 :-")
print(s1.name)
print(s1.roll_no)
print(s1.marks)


# Accessing the attributes of instance s2
print("Instance 2 :-")
print(s2.name)
print(s2.roll_no)
print(s2.marks)
```
