```python
# Creating a Student class
class student :

    # Constructor: runs automatically when an instance is created
    def __init__ (self , name , roll_no , marks) :

        # Normal instance attributes
        self.name = name 
        self.roll_no = roll_no

        # _marks is an encapsulated/protected-style attribute
        # It is intended to be accessed through getter and setter methods
        self._marks = marks

    # Getter method
    # Used to access/read the value of the _marks attribute
    def get_marks(self) :
        return self._marks


    # Setter method
    # Used to modify/change the value of the _marks attribute
    def set_marks(self , marks):

        # Allow marks to be changed only if they are greater than 0
        if(marks > 0) :
            self._marks = marks
```

```python
# Creating an instance of the Student class
s1 = student('kartik' , 60 , 92)

# Accessing normal instance attributes directly
print(s1.name)
print(s1.roll_no)


# Accessing the encapsulated _marks attribute through the getter method
# get_marks() returns the current value of _marks
s1.get_marks()
```

```python
# Trying to change marks using the setter method
# 'hello' is a string, so it cannot be compared with the integer 0.
# Therefore, Python raises a TypeError.
s1.set_marks('hello')
```

```python
# Setting a valid marks value using the setter
s1.set_marks(94)
```

```python
# Getting the updated marks using the getter
s1.get_marks()
```
