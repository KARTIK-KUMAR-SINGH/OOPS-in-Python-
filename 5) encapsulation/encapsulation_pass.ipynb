```python
# Creating a Student class
class student :

    # Constructor: runs automatically when an instance is created
    def __init__ (self , name , roll_no , marks) :

        # Normal instance attributes
        self.name = name 
        self.roll_no = roll_no

        # _marks is intended to be an internal/protected-style attribute
        # It is conventionally meant to be accessed through methods
        self._marks = marks
```

```python
# Creating an instance of the Student class
s1 = student('kartik' , 60 , 94)

# Accessing normal instance attributes
print(s1.name)
# Accessing the _marks attribute directly
# This bypasses the intended encapsulation.
# Python still allows this because _marks is NOT truly private.
print(s1._marks)
```
