```python
# Creating a Student class
class student :

    # Constructor: runs automatically when an instance is created
    def __init__ (self , name , roll_no , marks) :

        # Instance attributes
        self.name = name 
        self.roll_no = roll_no

        # _marks is an encapsulated/protected-style attribute
        # It is intended to be accessed through methods
        self._marks = marks
```

```python
# Creating two instances of the Student class
s1 = student('kartik' , 60 , 94)
s2 = student('abhishek' , 62 , 96)

# Accessing normal instance attributes
print(s1.name)
print(s1.roll_no)
# ERROR:
# The attribute was created as _marks, not marks.
# Therefore, s1.marks does not exist.
print(s1.marks)
```

```python
# Creating a Student class
class student :

    # Constructor: runs automatically when an instance is created
    def __init__ (self , name , roll_no , marks) :

        # These are public instance attributes.
        # They can be accessed directly from outside the class.
        self.name = name 
        self.roll_no = roll_no 
        self.marks = marks
```

```python
# Creating two instances of the Student class
s1 = student('kartik' , 60 , 94)
s2 = student('abhishek' , 62 , 96)


# Directly accessing the instance attributes
# No getter or setter method is required.
print(s1.name)
print(s1.roll_no)
print(s1.marks)

print(s2.name)
print(s2.roll_no)
print(s2.marks)
```
