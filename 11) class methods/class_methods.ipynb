```python
class student :
    # Class variable shared by all Student objects.
    student_count = 0

    def __init__(self , name , age) :
        # Instance variables belong to each individual object.
        self.name = name
        self.age = age

        # Increase the shared class variable whenever a new
        # Student object is created.
        student.student_count += 1

    # Instance method: normally called through an object.
    def introduce(self) :
        return f"The student name is :- {self.name} and his age is :- {self.age}"

    # Class method: can be called directly using the class.
    # It receives the class itself through 'cls'.
    @classmethod
    def get_student_count(cls) :
        return cls.student_count
```

```python
# Creating Student objects increases the shared student_count.
s1 = student("ajay" , 22)
s2 = student("abhishek" , 21)
```

```python
# Because get_student_count() is a class method,
# we can call it directly using the class name.
# No object is required to access this method.
student.get_student_count()
```
