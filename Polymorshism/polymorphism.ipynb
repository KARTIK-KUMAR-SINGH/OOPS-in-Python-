```python
# Parent class
class animal :
    def speak(self) :
        return "animal makes a sound"

# Child class 1
class dog(animal) :

    # Overriding the speak() method
    def speak(self) :
        return "bhaoo-bhaoo"

# Child class 2
class cat(animal) :

    # Overriding the speak() method
    def speak(self):
        return "meow-meow"

# Child class 3
class cow(animal) :

    # Overriding the speak() method
    def cow(self):
        return "moow-moow"
```

```python
# Creating objects of different classes.
# All three classes have a speak() method,
# but each class provides different behavior.
dog = dog()
cat = cat()
cow = cow()

# Calling the same method on different objects.
# This demonstrates polymorphism:
# the same method call produces different outputs
# depending on the object.
print(dog.speak()) # Dog's version of speak()
print(cat.speak()) # Cat's version of speak()
print(cow.speak()) # Cow's version of speak()
```
