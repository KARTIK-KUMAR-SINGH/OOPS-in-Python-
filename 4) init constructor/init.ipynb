```python
# Creating a Book class
class book :

    # __init__ is the constructor.
    # It is automatically called when we create an instance of the class.
    def __init__(self , author , title , price) :

        # These are instance attributes.
        # Each book instance will have its own author, title and price. 
        self.author = author 
        self.title = title
        self.price = price
```

```python
# Creating the first instance of the Book class
b1 = book("Atomic Habits" , "James Clear" , 500)

# Creating the second instance of the Book class
b2 = book("Rich Dad Poor Dad" , "Robert Kiyosaki" , 400)

# Accessing the attributes of the first instance
print(b1.author)
print(b1.title)
print(b1.price)

# Accessing the attributes of the second instance
print(b2.author)
print(b2.title)
print(b2.price)
```
