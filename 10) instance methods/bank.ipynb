```python
class bank :
    def __init__(self) :
        self.pin = ''
        self.balance = 0
        self.menu()

    def menu(self):
        user_input = input('''
        Hello, how would you like to proceed?
        1. Enter 1 to create pin
        2. Enter 2 to change pin
        3. Enter 3 to check balance
        4. Enter 4 to withdraw
        5. Press anythong to exit
        ''')

        if(user_input == '1'):
            #create new pin
            self.create_pin()
        elif(user_input == '2'):
            #change the pin
            self.change_pin()
            pass
        elif(user_input == '3'):
            #check the balanace
            self.check_balance()
        elif(user_input == '4'):
            #withdraw money
            self.withdraw_money()
            pass
        else :
            exit()

    def create_pin(self):
        user_pin = input("Enter the pin")
        self.pin = user_pin
        print("Pin created sucessfully")

        user_balance = int(input("Enter the balance"))
        self.balance = user_balance
        print("Balance added sucessfully")
        self.menu()
    def change_pin(self):
        old_pin = input("Enter your old pin")
        if(old_pin == self.pin):
            new_pin = input("Enter your new pin")
            self.pin = new_pin
            print("The pin is updated sucessfully")

        else :
            print("The entered pin is not correct")
        self.menu()

    def check_balance(self):
        check_pin = input("Enter your pin")
        if(check_pin == self.pin):
            print(f"your balance is {self.balance}")
        else:
            print("The entered pin is not correct")
        self.menu()

    def withdraw_money(self) :
        check_pin = input("Enter your pin")
        if(check_pin == self.pin):
            money = int(input("Enter the money to withdraw"))
            if(money > self.balance):
                print("Insufficent balance")
            else:
                self.balance = self.balance - money
                print("The money is withdraw sucessfully")
                print(f"Your current balance is :- {self.balance}")
        else:
            print("The entered pin is not correct")
        self.menu()
```
