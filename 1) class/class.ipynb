```python
# Creating a Student class
class student :

     # __init__ is the constructor.
    # It runs automatically when a new instance of the class is created.
    def __init__(self , name , roll_no , marks) :

        # These are instance attributes.
        # Each instance will have its own name, roll_no and marks.
        self.name = name 
        self.roll_no = roll_no
        self.marks = marks

    # This method checks whether the student has passed or not.
    def is_pass(self) :

        # If marks are 40 or above, the student has passed.
        if(self.marks >= 40) :
            return True
        else :
            return False

     # This method determines the grade of the student.
    def grade(self) :

        # Marks 80 or above -> Grade A
        if(self.marks >= 80):
            return "A"

        # Marks 60 or above -> Grade B
        elif(self.marks >= 60):
            return "B"

        # Marks 40 or above -> Grade C
        elif(self.marks >= 40):
            return "C"

        # Marks below 40 -> Grade F
        else:
            return "F"
```

```python
# Creating an instance/object of the Student class.
# 'kartik' is assigned to name,
# 53 is assigned to roll_no,
# and 100 is assigned to marks.
s1 = student('kartik' , 53 , 100)

# Calling the is_pass() method using the s1 instance.
# It returns True because marks are greater than or equal to 40.
s1.is_pass()

# Calling the grade() method using the s1 instance.
# It returns "A" because marks are greater than or equal to 80.
s1.grade()
```
